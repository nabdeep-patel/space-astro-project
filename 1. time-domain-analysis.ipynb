```python
import warnings
warnings.filterwarnings("ignore", "Wswiglal-redir-stdio")

import pandas as pd
import numpy as np
import gwpy
```

### For first Black Hole Merger of highly differnt masses GW190412 ($27.7 M_{\odot}$, $9.0 M_{\odot}$)

```python
data = pd.read_html("https://gwosc.org/eventapi/html/GWTC/?page=1&pagesize=100")

# Extract all events data
events = data[0]
merger = events
merger[merger["Name"]=='GW190412']
```

```python
from gwosc.datasets import event_gps

# first bbh merger
gps = event_gps('GW190412')
print(gps)
```

```python
segment = (int(gps)-5, int(gps)+5)
print(segment)
```

```python
from gwpy.timeseries import TimeSeries
ldata = TimeSeries.fetch_open_data('L1', *segment, verbose=True)
vdata = TimeSeries.fetch_open_data('V1', *segment, verbose=True)
hdata = TimeSeries.fetch_open_data('H1', *segment, verbose=True)
print(ldata)
```

```python
import matplotlib.pyplot as plt
plot = ldata.plot()
plot = vdata.plot()
plot = hdata.plot()
```
